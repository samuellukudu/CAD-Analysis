```python
"""
analyze_chunks.py

Updates:
1. Calculates Physical Size from GeoJSON Geometry (not normalized features).
2. Improved Structural Diversity metrics (Singleton vs. Repeated structures).
3. Added Average Degree to topology stats.
"""

import json
import numpy as np
import networkx as nx
import torch
from torch_geometric.data import Data
from torch_geometric.utils import to_networkx
from collections import Counter, defaultdict

# ==========================================
# 1. METRIC CALCULATORS
# ==========================================

def get_basic_stats(chunks):
    """Calculates basic node/edge counts and physical dimensions."""
    if not chunks: return {}
    
    node_counts = np.array([g.num_nodes for g in chunks])
    edge_counts = np.array([g.num_edges for g in chunks])
    
    # Calculate physical width/height from the pre-calculated bounds stored in loader
    widths = [g.phys_size[0] for g in chunks]
    heights = [g.phys_size[1] for g in chunks]
    
    return {
        'total_nodes': node_counts.sum(),
        'total_edges': edge_counts.sum(),
        'nodes_min': node_counts.min(),
        'nodes_max': node_counts.max(),
        'nodes_avg': node_counts.mean(),
        'width_avg': np.mean(widths) if widths else 0.0,
        'height_avg': np.mean(heights) if heights else 0.0,
        'width_max': np.max(widths) if widths else 0.0,
        'height_max': np.max(heights) if heights else 0.0
    }

def get_advanced_topology(chunks):
    """Calculates Graph Theory metrics."""
    diameters = []
    clustering_coeffs = []
    articulation_points = []
    avg_degrees = []
    
    for g in chunks:
        # Convert to NetworkX (Undirected)
        nx_g = to_networkx(g, to_undirected=True)
        n = nx_g.number_of_nodes()
        
        # Average Degree
        if n > 0:
            degrees = [d for n, d in nx_g.degree()]
            avg_degrees.append(sum(degrees) / n)
        else:
            avg_degrees.append(0)

        # Diameter (Handle disconnected graphs by taking max of components)
        if n > 1:
            if nx.is_connected(nx_g):
                diameters.append(nx.diameter(nx_g))
            else:
                comps = list(nx.connected_components(nx_g))
                if comps:
                    largest = nx_g.subgraph(max(comps, key=len))
                    diameters.append(nx.diameter(largest))
                else:
                    diameters.append(0)
        else:
            diameters.append(0)
            
        # Clustering & Critical Nodes
        clustering_coeffs.append(nx.average_clustering(nx_g))
        articulation_points.append(len(list(nx.articulation_points(nx_g))))

    return {
        'avg_degree': np.mean(avg_degrees),
        'avg_diameter': np.mean(diameters),
        'avg_clustering': np.mean(clustering_coeffs),
        'avg_critical_nodes': np.mean(articulation_points)
    }

def get_diversity_stats(chunks):
    """Analyzes the structure_id labels."""
    labels = []
    if chunks and chunks[0].y is not None:
        for c in chunks: labels.extend(c.y.tolist())
    
    if not labels: return None

    counts = Counter(labels)
    total_items = len(labels)
    unique_types = len(counts)
    
    # "Repeated Structures": IDs that appear more than once
    repeated_types = {k: v for k, v in counts.items() if v > 1}
    num_repeated_nodes = sum(repeated_types.values())
    
    return {
        'unique_types': unique_types,
        'most_common_id': counts.most_common(1)[0],
        'singleton_count': unique_types - len(repeated_types),
        'repetition_ratio': num_repeated_nodes / total_items # % of nodes that are parts of recurring patterns
    }

# ==========================================
# 2. DATA LOADER
# ==========================================

def load_dataset_by_layout(geojson_path):
    print(f"[INFO] Loading {geojson_path}...")
    try:
        with open(geojson_path, 'r') as f:
            data = json.load(f)
    except FileNotFoundError:
        return {}

    # Group features by Layout -> Chunk
    grouped_props = defaultdict(lambda: defaultdict(list))
    grouped_geoms = defaultdict(lambda: defaultdict(list))

    for feature in data['features']:
        p = feature['properties']
        lid, cid = p['layout_id'], p['chunk_id']
        grouped_props[lid][cid].append(p)
        grouped_geoms[lid][cid].append(feature['geometry']['coordinates'])

    layout_graphs = {}

    print(f"[INFO] Constructing PyG graphs...")
    for lid in sorted(grouped_props.keys()):
        chunk_list = []
        chunks_dict = grouped_props[lid]
        
        for cid in sorted(chunks_dict.keys()):
            node_props = chunks_dict[cid]
            node_geoms = grouped_geoms[lid][cid]
            
            # Sort by node_id to ensure adjacency matrix aligns
            # Zip them to keep geometry aligned with properties
            combined = sorted(zip(node_props, node_geoms), key=lambda x: x[0]['node_id'])
            sorted_props = [x[0] for x in combined]
            sorted_geoms = [x[1] for x in combined]
            
            # 1. Tensor Features
            feats = [n['features'] for n in sorted_props]
            x = torch.tensor(np.array(feats), dtype=torch.float)
            
            # 2. Tensor Labels (Structure ID)
            y = None
            if 'structure_id' in sorted_props[0]:
                y = torch.tensor([n['structure_id'] for n in sorted_props], dtype=torch.long)

            # 3. Adjacency (Edge Index)
            edges = []
            for n in sorted_props:
                u = int(n['node_id'])
                for v in n['neighbors']:
                    edges.append([u, int(v)])
            
            edge_index = torch.tensor(edges, dtype=torch.long).t().contiguous() if edges else torch.empty((2, 0), dtype=torch.long)

            # 4. Physical Size Calculation (From Raw Geometry)
            # Flatten all points in this chunk to find bounding box
            all_x, all_y = [], []
            for coords in sorted_geoms:
                # coords is usually [[[x,y], [x,y]...]] (Polygon)
                # We flatten simply to get min/max
                pts = np.array(coords).reshape(-1, 2) 
                all_x.extend(pts[:, 0])
                all_y.extend(pts[:, 1])
            
            phys_w = max(all_x) - min(all_x) if all_x else 0
            phys_h = max(all_y) - min(all_y) if all_y else 0

            # 5. Build Object
            d = Data(x=x, edge_index=edge_index, y=y)
            d.layout_id = lid
            d.chunk_id = cid
            d.num_nodes = x.shape[0]
            d.phys_size = (phys_w, phys_h) # Store raw size tuple
            
            chunk_list.append(d)
            
        layout_graphs[lid] = chunk_list

    return layout_graphs

# ==========================================
# 3. REPORT GENERATOR
# ==========================================

def print_layout_report(layout_id, chunks):
    basic = get_basic_stats(chunks)
    topo = get_advanced_topology(chunks)
    div = get_diversity_stats(chunks)
    
    print(f"\n" + "="*60)
    print(f" LAYOUT {layout_id} REPORT")
    print("="*60)
    
    print(f"Contains {len(chunks)} Chunks | {basic['total_nodes']} Total Nodes")
    
    # Table 1: Size
    print(f"\n[A] GEOMETRY & SCALE")
    print(f" - Nodes per Chunk:      Min {basic['nodes_min']} | Avg {basic['nodes_avg']:.1f} | Max {basic['nodes_max']}")
    print(f" - Physical Dim (Avg):   {basic['width_avg']:.1f} W  x {basic['height_avg']:.1f} H (Units)")
    print(f" - Largest Chunk:        {basic['width_max']:.1f} W  x {basic['height_max']:.1f} H")
    
    # Table 2: Topology
    print(f"\n[B] GRAPH TOPOLOGY")
    print(f" - Avg Degree:           {topo['avg_degree']:.1f} neighbors per node")
    print(f" - Avg Diameter:         {topo['avg_diameter']:.1f} steps (Max distance)")
    print(f" - Clustering Coeff:     {topo['avg_clustering']:.3f}")
    print(f"   (>0.5 indicates dense, mesh-like connectivity)")
    print(f" - Critical Points:      {topo['avg_critical_nodes']:.1f} per chunk")

    # Table 3: Diversity
    if div:
        print(f"\n[C] STRUCTURAL DIVERSITY")
        print(f" - Unique Structures:    {div['unique_types']} types found")
        print(f" - Pattern Reuse:        {div['repetition_ratio']*100:.1f}% of nodes are part of repeated patterns")
        print(f" - Singletons:           {div['singleton_count']} (Unique shapes appearing only once)")
        top_id, top_count = div['most_common_id']
        print(f" - Most Common:          ID {top_id} (Appears {top_count} times)")
    else:
        print(f"\n[C] DIVERSITY: No structural labels found.")

# ==========================================
# MAIN
# ==========================================
if __name__ == "__main__":
    file_path = "chunked_data/AEC Plan Elev Sample.geojson"
    
    layouts = load_dataset_by_layout(file_path)
    
    if not layouts:
        print(f"Error: Could not load {file_path}")
    else:
        print(f"\n[INFO] Dataset contains {len(layouts)} unique Layouts.")
        for lid, chunks in layouts.items():
            print_layout_report(lid, chunks)
```

```python
# import os
# import json
# import torch
# import numpy as np
# from torch_geometric.data import InMemoryDataset, Data
# from collections import defaultdict
# from shapely.geometry import Polygon
# import shutil

# class ArchitecturalGraphDataset(InMemoryDataset):
#     def __init__(self, root, source_path, transform=None, pre_transform=None, force_reprocess=False):
#         self.source_path = source_path
        
#         # 1. Determine Input Type
#         self._file_list = []
#         if os.path.isdir(self.source_path):
#             self._file_list = sorted([f for f in os.listdir(self.source_path) if f.endswith('.geojson')])
#             self.input_name = "unified_folder"
#         elif os.path.isfile(self.source_path):
#             self._file_list = [os.path.basename(self.source_path)]
#             # Use filename without extension as the unique cache ID
#             self.input_name = os.path.splitext(os.path.basename(self.source_path))[0]
#         else:
#             raise FileNotFoundError(f"Source path not found: {self.source_path}")

#         if not self._file_list:
#             raise RuntimeError(f"No .geojson files found in {self.source_path}")

#         # Handle Force Reprocess
#         if force_reprocess:
#             # We only delete the SPECIFIC processed file for this input, not the whole folder
#             processed_path = os.path.join(root, 'processed', self.processed_file_names[0])
#             if os.path.exists(processed_path):
#                 print(f"[INFO] force_reprocess=True. Deleting {processed_path}...")
#                 os.remove(processed_path)

#         super().__init__(root, transform, pre_transform)
#         self.data, self.slices = torch.load(self.processed_paths[0], weights_only=False)

#     @property
#     def raw_file_names(self):
#         return self._file_list

#     @property
#     def processed_file_names(self):
#         # [FIX] The cache filename now depends on the input name
#         # file_A.geojson -> data_file_A.pt
#         # folder/ -> data_unified_folder.pt
#         return [f'data_{self.input_name}.pt']

#     def download(self):
#         # Copy logic
#         for file_name in self.raw_file_names:
#             src = os.path.join(self.source_path, file_name) if os.path.isdir(self.source_path) else self.source_path
#             dst = os.path.join(self.raw_dir, file_name)
#             if not os.path.exists(dst):
#                 shutil.copy(src, dst)

#     def process(self):
#         print(f"[INFO] Generating {self.processed_file_names[0]} from {len(self.raw_paths)} files...")
#         data_list = []
        
#         for file_idx, raw_path in enumerate(self.raw_paths):
#             print(f"   -> [{file_idx+1}/{len(self.raw_paths)}] Loading {os.path.basename(raw_path)}")
            
#             with open(raw_path, 'r') as f:
#                 data_json = json.load(f)

#             grouped_data = defaultdict(lambda: defaultdict(list))
            
#             # Parse
#             for feature in data_json['features']:
#                 p = feature['properties']
#                 geom = feature['geometry']['coordinates']
#                 grouped_data[p['layout_id']][p['chunk_id']].append((p, geom))
                
#             # Build Graphs
#             for lid in sorted(grouped_data.keys()):
#                 for cid in sorted(grouped_data[lid].keys()):
#                     items = grouped_data[lid][cid]
#                     items.sort(key=lambda x: x[0]['node_id'])
                    
#                     node_props = [i[0] for i in items]
#                     raw_coords = [i[1] for i in items]

#                     features = [n['features'] for n in node_props]
#                     x = torch.tensor(np.array(features), dtype=torch.float)
                    
#                     if 'structure_id' in node_props[0]:
#                         y = torch.tensor([n['structure_id'] for n in node_props], dtype=torch.long)
#                     else:
#                         y = None

#                     edge_list = []
#                     for n in node_props:
#                         u = int(n['node_id'])
#                         if 'neighbors' in n:
#                             for v in n['neighbors']:
#                                 edge_list.append([u, int(v)])

#                     if edge_list:
#                         edge_index = torch.tensor(edge_list, dtype=torch.long).t().contiguous()
#                     else:
#                         edge_index = torch.empty((2, 0), dtype=torch.long)

#                     centroids = []
#                     for coords in raw_coords:
#                         if len(coords) > 0:
#                             poly = Polygon(coords[0])
#                             centroids.append([poly.centroid.x, poly.centroid.y])
#                         else:
#                             centroids.append([0.0, 0.0])
#                     pos = torch.tensor(np.array(centroids), dtype=torch.float)

#                     data = Data(x=x, edge_index=edge_index, y=y, pos=pos)
                    
#                     # Metadata
#                     data.layout_id = torch.tensor([int(lid)], dtype=torch.long)
#                     data.chunk_id = torch.tensor([int(cid)], dtype=torch.long)
#                     data.file_id = torch.tensor([int(file_idx)], dtype=torch.long)
                    
#                     # Store file name hash/string to verify matching later if needed
#                     # data.filename = os.path.basename(raw_path) 

#                     if self.pre_transform is not None:
#                         data = self.pre_transform(data)

#                     data_list.append(data)

#         data, slices = self.collate(data_list)
#         print(f"[SUCCESS] Saved {len(data_list)} graphs to {self.processed_paths[0]}")
#         torch.save((data, slices), self.processed_paths[0])

import os
import shutil
import json
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import random
from collections import defaultdict
from shapely.geometry import Polygon
from torch_geometric.data import InMemoryDataset, Data
from torch_geometric.loader import DataLoader
from torch_geometric.nn import GCNConv, global_mean_pool
from torch_geometric.utils import get_laplacian, to_dense_adj, dropout_edge

# ==========================================
# 1. DATA AUGMENTATION
# ==========================================
def augment_batch(batch, noise_level=0.02, edge_drop_prob=0.1):
    batch_aug = batch.clone()
    if batch_aug.x is not None:
        noise = torch.randn_like(batch_aug.x) * noise_level
        batch_aug.x += noise
    if batch_aug.edge_index.numel() > 0:
        edge_index, _ = dropout_edge(batch_aug.edge_index, p=edge_drop_prob)
        batch_aug.edge_index = edge_index
    return batch_aug

# ==========================================
# 2. DATASET CLASS
# ==========================================
class ArchitecturalGraphDataset(InMemoryDataset):
    def __init__(self, root, source_path, transform=None, pre_transform=None, force_reprocess=False):
        self.source_path = source_path
        
        # 1. Determine Input Type & Cache Name
        self._file_list = []
        if os.path.isdir(self.source_path):
            self._file_list = sorted([f for f in os.listdir(self.source_path) if f.endswith('.geojson')])
            self.input_name = "unified_folder"
        elif os.path.isfile(self.source_path):
            self._file_list = [os.path.basename(self.source_path)]
            self.input_name = os.path.splitext(os.path.basename(self.source_path))[0]
        else:
            raise FileNotFoundError(f"Source path not found: {self.source_path}")

        if not self._file_list:
            raise RuntimeError(f"No .geojson files found in {self.source_path}")

        # Handle Force Reprocess
        if force_reprocess:
            processed_path = os.path.join(root, 'processed', self.processed_file_names[0])
            if os.path.exists(processed_path):
                print(f"[INFO] force_reprocess=True. Deleting {processed_path}...")
                os.remove(processed_path)

        super().__init__(root, transform, pre_transform)
        self.data, self.slices = torch.load(self.processed_paths[0], weights_only=False)

    @property
    def raw_file_names(self):
        return self._file_list

    @property
    def processed_file_names(self):
        # Dynamic cache name based on input
        return [f'data_{self.input_name}.pt']

    def download(self):
        for file_name in self.raw_file_names:
            src = os.path.join(self.source_path, file_name) if os.path.isdir(self.source_path) else self.source_path
            dst = os.path.join(self.raw_dir, file_name)
            if not os.path.exists(dst):
                shutil.copy(src, dst)

    def process(self):
        print(f"[INFO] Generating {self.processed_file_names[0]} from {len(self.raw_paths)} files...")
        data_list = []
        
        for file_idx, raw_path in enumerate(self.raw_paths):
            print(f"   -> [{file_idx+1}/{len(self.raw_paths)}] Loading {os.path.basename(raw_path)}")
            
            with open(raw_path, 'r') as f:
                data_json = json.load(f)

            grouped_data = defaultdict(lambda: defaultdict(list))
            for feature in data_json['features']:
                p = feature['properties']
                geom = feature['geometry']['coordinates']
                grouped_data[p['layout_id']][p['chunk_id']].append((p, geom))
                
            for lid in sorted(grouped_data.keys()):
                for cid in sorted(grouped_data[lid].keys()):
                    items = grouped_data[lid][cid]
                    items.sort(key=lambda x: x[0]['node_id'])
                    
                    node_props = [i[0] for i in items]
                    raw_coords = [i[1] for i in items]

                    features = [n['features'] for n in node_props]
                    x = torch.tensor(np.array(features), dtype=torch.float)
                    y = torch.tensor([n.get('structure_id', 0) for n in node_props], dtype=torch.long)

                    edge_list = []
                    for n in node_props:
                        u = int(n['node_id'])
                        if 'neighbors' in n:
                            for v in n['neighbors']:
                                edge_list.append([u, int(v)])

                    if edge_list:
                        edge_index = torch.tensor(edge_list, dtype=torch.long).t().contiguous()
                    else:
                        edge_index = torch.empty((2, 0), dtype=torch.long)

                    centroids = []
                    for coords in raw_coords:
                        if len(coords) > 0:
                            poly = Polygon(coords[0])
                            centroids.append([poly.centroid.x, poly.centroid.y])
                        else:
                            centroids.append([0.0, 0.0])
                    pos = torch.tensor(np.array(centroids), dtype=torch.float)
                    
                    # Laplacian PE
                    k = 8
                    num_nodes = x.size(0)
                    if num_nodes > 1 and edge_index.numel() > 0:
                        edge_index_lap, edge_weight_lap = get_laplacian(edge_index, normalization='sym', num_nodes=num_nodes)
                        L = to_dense_adj(edge_index_lap, edge_attr=edge_weight_lap, max_num_nodes=num_nodes)[0]
                        E, V = torch.linalg.eigh(L)
                        pe = V[:, 1 : min(num_nodes, k) + 1]
                        if pe.size(1) < k: pe = F.pad(pe, (0, k - pe.size(1)), value=0)
                    else:
                        pe = torch.zeros((num_nodes, k))

                    data = Data(x=x, edge_index=edge_index, y=y, pos=pos, pos_enc=pe.float())
                    
                    data.layout_id = torch.tensor([int(lid)], dtype=torch.long)
                    data.chunk_id = torch.tensor([int(cid)], dtype=torch.long)
                    data.file_id = torch.tensor([int(file_idx)], dtype=torch.long)
                    
                    # [CRITICAL] Global ID for Triplet Mining
                    data.global_layout_id = torch.tensor([file_idx * 10000 + int(lid)], dtype=torch.long)

                    if self.pre_transform is not None:
                        data = self.pre_transform(data)

                    data_list.append(data)

        data, slices = self.collate(data_list)
        print(f"[SUCCESS] Saved {len(data_list)} graphs.")
        torch.save((data, slices), self.processed_paths[0])
```

```python
import os
import json
import torch
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon as MplPolygon
from matplotlib.collections import PatchCollection, LineCollection
from collections import defaultdict
import random

# Import your dataset class (or ensure it's pasted in the same file)
# from dataset import ArchitecturalGraphDataset 

# ==========================================
# 1. HELPER: GEOMETRY LOOKUP (Supports Folders)
# ==========================================
def build_geometry_lookup(source_path):
    """
    Creates a lookup: (file_idx, layout_id, chunk_id) -> List of Polygon Coords
    Iterates through all files in source_path to match Dataset structure.
    """
    print(f"[INFO] Indexing GeoJSON geometry from {source_path}...")
    
    # 1. Determine file list (MUST match Dataset class sorting)
    if os.path.isdir(source_path):
        file_list = sorted([f for f in os.listdir(source_path) if f.endswith('.geojson')])
        base_dir = source_path
    elif os.path.isfile(source_path):
        file_list = [os.path.basename(source_path)]
        base_dir = os.path.dirname(source_path)
    else:
        raise FileNotFoundError(f"Path not found: {source_path}")

    # 2. Build Lookup
    # Key: (file_idx, layout_id, chunk_id) -> Value: [List of Coords]
    full_lookup = {} 
    
    for file_idx, file_name in enumerate(file_list):
        full_path = os.path.join(base_dir, file_name)
        
        with open(full_path, 'r') as f:
            data = json.load(f)
            
            # Temporary grouping for this file
            file_grouped = defaultdict(list)
            
            for feature in data['features']:
                props = feature['properties']
                lid = props['layout_id']
                cid = props['chunk_id']
                # Handle potential naming variations
                nid = props.get('node_id', props.get('chunk_node_id'))
                
                geom = feature['geometry']
                if geom and 'coordinates' in geom:
                    file_grouped[(lid, cid)].append((nid, geom['coordinates'][0]))
            
            # Sort by node_id and store in global lookup with file_idx
            for (lid, cid), items in file_grouped.items():
                items.sort(key=lambda x: x[0]) 
                coords_sorted = [item[1] for item in items]
                
                # UNIQUE KEY: (file_idx, layout_id, chunk_id)
                full_lookup[(file_idx, lid, cid)] = coords_sorted
                
    print(f"[INFO] Indexed geometries for {len(file_list)} files.")
    return full_lookup

def group_indices_by_global_layout(dataset):
    """
    Returns a dict: { global_layout_id: [dataset_index_1, dataset_index_2, ...] }
    """
    print("[INFO] Grouping dataset indices by Global Layout ID...")
    layout_map = defaultdict(list)
    
    # Access internal storage for speed
    if hasattr(dataset, '_data') and hasattr(dataset._data, 'global_layout_id'):
        global_ids = dataset._data.global_layout_id.numpy()
        for idx, gid in enumerate(global_ids):
            layout_map[gid].append(idx)
    else:
        # Fallback for older dataset versions (single file)
        for idx, data in enumerate(dataset):
            lid = data.layout_id.item()
            layout_map[lid].append(idx)
            
    return layout_map

# ==========================================
# 2. VISUALIZATION LOGIC
# ==========================================
def visualize_layouts(dataset, source_path, samples_per_layout=3, max_layouts_to_show=5):
    
    # 1. Prepare Data Maps
    geo_lookup = build_geometry_lookup(source_path)
    layout_indices = group_indices_by_global_layout(dataset)
    
    # Filter to layouts that actually have enough chunks to be interesting
    valid_layouts = [k for k, v in layout_indices.items() if len(v) >= 1]
    unique_layouts = sorted(valid_layouts)
    
    print(f"[INFO] Found {len(unique_layouts)} unique global layouts.")
    
    # Pick a few random layouts to visualize
    layouts_to_plot = random.sample(unique_layouts, min(len(unique_layouts), max_layouts_to_show))

    # 2. Iterate over Selected Layouts
    for gid in layouts_to_plot:
        available_indices = layout_indices[gid]
        
        # Determine how many to plot
        n_samples = min(len(available_indices), samples_per_layout)
        selected_indices = np.random.choice(available_indices, n_samples, replace=False)
        
        print(f"\n--- Visualizing Global Layout {gid} ({n_samples} chunks) ---")

        # Create Figure
        fig, axes = plt.subplots(n_samples, 2, figsize=(16, 6 * n_samples))
        if n_samples == 1: axes = np.array([axes]) # Ensure 2D array structure
        
        fig.suptitle(f"Global Layout ID: {gid}", fontsize=20, fontweight='bold', y=0.98)

        for row, idx in enumerate(selected_indices):
            data = dataset[idx]
            
            # --- INFO EXTRACTION ---
            # We need these 3 keys to find the geometry
            lid = data.layout_id.item()
            cid = data.chunk_id.item()
            
            # Handle File ID (Default to 0 if using old dataset)
            fid = data.file_id.item() if hasattr(data, 'file_id') else 0
            
            # --- A. Retrieve Geometry ---
            # Lookup Key: (file_idx, layout_id, chunk_id)
            poly_coords_list = geo_lookup.get((fid, lid, cid), [])
            
            polygons = []
            true_centroids = []
            all_coords = []
            
            for coords in poly_coords_list:
                all_coords.append(coords)
                polygons.append(MplPolygon(np.array(coords), closed=True))
                
                # Calculate Centroid
                pts = np.array(coords[:-1]) 
                if len(pts) > 0:
                    true_centroids.append(np.mean(pts, axis=0))
                else:
                    true_centroids.append([0,0])

            true_centroids = np.array(true_centroids)
            
            # --- B. Plot Left: Raw Geometry ---
            ax_geo = axes[row][0]
            
            if polygons:
                colors = plt.cm.tab20(np.linspace(0, 1, len(polygons)))
                pc = PatchCollection(polygons, facecolor=colors, edgecolor='black', alpha=0.7)
                ax_geo.add_collection(pc)
                
                # Dynamic Scaling
                if len(all_coords) > 0:
                    all_pts = np.concatenate([np.array(p) for p in all_coords])
                    min_x, min_y = all_pts.min(axis=0)
                    max_x, max_y = all_pts.max(axis=0)
                    pad = max(max_x-min_x, max_y-min_y) * 0.1
                    ax_geo.set_xlim(min_x - pad, max_x + pad)
                    ax_geo.set_ylim(min_y - pad, max_y + pad)
            
            ax_geo.set_title(f"File {fid} | Layout {lid} | Chunk {cid}\nGeometry", fontsize=12)
            ax_geo.set_aspect('equal')
            ax_geo.axis('off')

            # --- C. Plot Right: Graph Overlay ---
            ax_graph = axes[row][1]
            
            # 1. Draw Nodes
            if len(true_centroids) > 0:
                ax_graph.scatter(true_centroids[:, 0], true_centroids[:, 1], c='black', s=25, zorder=5)
                
                # 2. Draw Edges
                if data.edge_index is not None and data.edge_index.shape[1] > 0:
                    edge_index = data.edge_index.numpy()
                    lines = []
                    for e in range(edge_index.shape[1]):
                        u, v = edge_index[0, e], edge_index[1, e]
                        if u < len(true_centroids) and v < len(true_centroids):
                            lines.append([true_centroids[u], true_centroids[v]])
                    
                    if lines:
                        lc = LineCollection(lines, colors='red', linewidths=1.5, alpha=0.6, zorder=10)
                        ax_graph.add_collection(lc)

            # 3. Draw Ghost Background
            if polygons:
                pc_ghost = PatchCollection(polygons, facecolor='#eeeeee', edgecolor='#d0d0d0', alpha=0.5, zorder=0)
                ax_graph.add_collection(pc_ghost)

            # Sync Axis
            ax_graph.set_xlim(ax_geo.get_xlim())
            ax_graph.set_ylim(ax_geo.get_ylim())
            ax_graph.set_title("Graph Topology", fontsize=12)
            ax_graph.set_aspect('equal')
            ax_graph.axis('off')

        plt.tight_layout(rect=[0, 0, 1, 0.96]) 
        plt.show()

# ==========================================
# EXECUTION
# ==========================================
if __name__ == "__main__":
    
    # 1. SETUP
    ROOT_DIR = "./dataset_root"
    SOURCE_PATH = "chunked_data/"  # Folder or File
    
    # 2. Check Cache Logic (Same as train.py)
    processed_file = os.path.join(ROOT_DIR, "processed", "architectural_graphs_unified.pt")
    force_update = False
    
    if not os.path.exists(processed_file):
        force_update = True
    else:
        proc_mtime = os.path.getmtime(processed_file)
        if os.path.isdir(SOURCE_PATH):
            for f in os.listdir(SOURCE_PATH):
                if f.endswith('.geojson'):
                    if os.path.getmtime(os.path.join(SOURCE_PATH, f)) > proc_mtime:
                        force_update = True
                        break
        elif os.path.isfile(SOURCE_PATH):
             if os.path.getmtime(SOURCE_PATH) > proc_mtime:
                force_update = True

    # 3. Load Dataset
    # Assuming ArchitecturalGraphDataset is defined or imported
    dataset = ArchitecturalGraphDataset(
        root=ROOT_DIR, 
        source_path=SOURCE_PATH, 
        force_reprocess=force_update
    )

    # 4. Run Visualization
    # Will pick 5 random distinct buildings and show 3 chunks from each
    visualize_layouts(dataset, SOURCE_PATH, samples_per_layout=5, max_layouts_to_show=10)
```

```python
import os
import shutil
import json
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import random
from collections import defaultdict
from shapely.geometry import Polygon
from torch_geometric.data import InMemoryDataset, Data
from torch_geometric.loader import DataLoader
from torch_geometric.nn import GCNConv, global_mean_pool
from torch_geometric.utils import get_laplacian, to_dense_adj, dropout_edge

# ==========================================
# 3. MODEL
# ==========================================
class TrainedGraphEncoder(nn.Module):
    def __init__(self, in_channels, hidden_dim=128, embedding_dim=64):
        super().__init__()
        torch.manual_seed(42)
        self.conv1 = GCNConv(in_channels, hidden_dim)
        self.conv2 = GCNConv(hidden_dim, hidden_dim)
        self.conv3 = GCNConv(hidden_dim, hidden_dim)
        self.lin = nn.Linear(hidden_dim, embedding_dim)

    def forward(self, x, edge_index, batch):
        x = self.conv1(x, edge_index).relu()
        x = F.dropout(x, p=0.1, training=self.training)
        x = self.conv2(x, edge_index).relu()
        x = self.conv3(x, edge_index)
        x = global_mean_pool(x, batch)
        x = self.lin(x)
        x = F.normalize(x, p=2, dim=1) 
        return x

# ==========================================
# 1. ROBUST SPLIT STRATEGY
# ==========================================
def split_by_layout(dataset, val_ratio=0.2):
    layout_map = defaultdict(list)
    
    if not hasattr(dataset._data, 'global_layout_id'):
        raise AttributeError("Dataset missing 'global_layout_id'. Rebuild cache.")

    global_ids = dataset._data.global_layout_id.numpy()
    
    for idx, gid in enumerate(global_ids):
        layout_map[gid].append(idx)
        
    valid_layouts = []
    
    # Filter Singletons
    for gid, indices in layout_map.items():
        if len(indices) >= 2:
            valid_layouts.append(gid)

    if len(valid_layouts) < 2:
        raise RuntimeError("Not enough valid layouts (>=2 chunks) to split!")

    random.seed(42)
    random.shuffle(valid_layouts)
    
    n_val = int(len(valid_layouts) * val_ratio)
    if n_val == 0 and len(valid_layouts) > 1: n_val = 1
        
    val_layout_ids = set(valid_layouts[:n_val])
    train_indices, val_indices = [], []
    
    for gid in valid_layouts:
        indices = layout_map[gid]
        if gid in val_layout_ids:
            val_indices.extend(indices)
        else:
            train_indices.extend(indices)
            
    print(f"[SPLIT] Train: {len(train_indices)} graphs | Val: {len(val_indices)} graphs")
    return dataset[train_indices], dataset[val_indices]

# ==========================================
# 2. AUGMENTATION
# ==========================================
def augment_batch(batch, noise_level=0.02, edge_drop_prob=0.1):
    batch_aug = batch.clone()
    if batch_aug.x is not None:
        noise = torch.randn_like(batch_aug.x) * noise_level
        batch_aug.x += noise
    if batch_aug.edge_index.numel() > 0:
        edge_index, _ = dropout_edge(batch_aug.edge_index, p=edge_drop_prob)
        batch_aug.edge_index = edge_index
    return batch_aug

# ==========================================
# 3. TRAINING ENGINE
# ==========================================
def run_epoch(loader, model, criterion, optimizer, device, is_training=True):
    if is_training: model.train()
    else: model.eval()

    total_loss = 0.0
    valid_batches = 0

    for batch in loader:
        batch = batch.to(device)

        if is_training:
            batch = augment_batch(batch)
            optimizer.zero_grad()

        embeddings = model(batch.x, batch.edge_index, batch.batch)
        
        labels = batch.global_layout_id.view(-1)
        
        # Masks
        is_pos = labels.unsqueeze(0) == labels.unsqueeze(1)
        is_neg = ~is_pos
        is_pos.fill_diagonal_(False)

        triplet_loss = torch.tensor(0.0, device=device)
        num_triplets = 0
        
        for i in range(len(embeddings)):
            pos_indices = is_pos[i].nonzero(as_tuple=True)[0]
            neg_indices = is_neg[i].nonzero(as_tuple=True)[0]
            
            if len(pos_indices) > 0 and len(neg_indices) > 0:
                anchor = embeddings[i]
                positive = embeddings[pos_indices[random.randint(0, len(pos_indices)-1)]]
                negative = embeddings[neg_indices[random.randint(0, len(neg_indices)-1)]]
                
                triplet_loss += criterion(anchor, positive, negative)
                num_triplets += 1
        
        if num_triplets > 0:
            triplet_loss = triplet_loss / num_triplets
            
            if is_training:
                triplet_loss.backward()
                
                # [NEW] Gradient Clipping: Prevents massive spikes
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                
                optimizer.step()
                
            total_loss += triplet_loss.item()
            valid_batches += 1

    if valid_batches == 0: return None
    return total_loss / valid_batches

# ==========================================
# 4. MAIN
# ==========================================
def main():
    ROOT_DIR = "./dataset_root"
    SOURCE_PATH = "chunked_data"
    
    # Increased epochs slightly because SGD takes longer to converge than Adam
    EPOCHS = 150 
    BATCH_SIZE = 32
    DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    print(f"[INFO] Device: {DEVICE}")

    # 1. Dataset
    if os.path.isdir(SOURCE_PATH): expected_name = "data_unified_folder.pt"
    else: base = os.path.splitext(os.path.basename(SOURCE_PATH))[0]; expected_name = f"data_{base}.pt"

    processed_file = os.path.join(ROOT_DIR, "processed", expected_name)
    force_update = not os.path.exists(processed_file)
    
    dataset = ArchitecturalGraphDataset(ROOT_DIR, SOURCE_PATH, force_reprocess=force_update)
    
    if not hasattr(dataset._data, 'global_layout_id'):
        print("[WARN] Rebuilding stale cache...")
        dataset = ArchitecturalGraphDataset(ROOT_DIR, SOURCE_PATH, force_reprocess=True)

    # 2. Split
    train_dataset, val_dataset = split_by_layout(dataset, val_ratio=0.2)

    # 3. Loaders
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
    val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=False)

    # 4. Model & Optimizer
    model = TrainedGraphEncoder(in_channels=dataset.num_features).to(DEVICE)
    
    # [NEW] SGD + Momentum + Weight Decay (Regularization)
    optimizer = torch.optim.SGD(model.parameters(), lr=0.01, momentum=0.9, weight_decay=1e-4)
    
    # [NEW] Learning Rate Scheduler
    # Decays LR by half every 25 epochs
    scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=25, gamma=0.5)

    criterion = nn.TripletMarginLoss(margin=1.0, p=2)

    best_val_loss = float('inf')
    
    print("\n[INFO] Training with SGD + Momentum...")
    for epoch in range(1, EPOCHS + 1):
        
        train_loss = run_epoch(train_loader, model, criterion, optimizer, DEVICE, is_training=True)
        
        val_loss = None
        if len(val_loader) > 0:
            val_loss = run_epoch(val_loader, model, criterion, optimizer, DEVICE, is_training=False)
        
        # Step the scheduler
        scheduler.step()
        current_lr = scheduler.get_last_lr()[0]

        t_str = f"{train_loss:.4f}" if train_loss is not None else "nan"
        v_str = f"{val_loss:.4f}" if val_loss is not None else "nan"
        
        if epoch % 5 == 0 or epoch == 1:
            print(f"Epoch {epoch:03d} | LR: {current_lr:.5f} | Train: {t_str} | Val: {v_str}")
            
        if val_loss is not None and val_loss < best_val_loss and val_loss > 1e-6:
            best_val_loss = val_loss
            torch.save(model.state_dict(), "trained_model.pth")

    print(f"[DONE] Best Valid Val Loss: {best_val_loss:.4f}")

if __name__ == "__main__":
    main()
```
