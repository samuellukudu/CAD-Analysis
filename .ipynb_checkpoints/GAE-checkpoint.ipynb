# Graph Autoencoders

```python
import json
import torch
import numpy as np
from collections import defaultdict
from torch_geometric.data import Data, InMemoryDataset
from torch_geometric.loader import DataLoader
import torch_geometric.transforms as T
import os
import shutil

class ArchitecturalGraphDataset(InMemoryDataset):
    def __init__(self, root, geojson_path, transform=None, pre_transform=None):
        self.geojson_path = geojson_path
        super().__init__(root, transform, pre_transform)
        self.data, self.slices = torch.load(self.processed_paths[0], weights_only=False)

    @property
    def raw_file_names(self):
        return [self.geojson_path]

    @property
    def processed_file_names(self):
        return ['architectural_graphs.pt']

    def process(self):
        print(f"[INFO] Processing {self.geojson_path}...")
        
        with open(self.geojson_path, 'r') as f:
            geojson = json.load(f)

        grouped_data = defaultdict(list)
        for feature in geojson['features']:
            props = feature['properties']
            grouped_data[(props['layout_id'], props['chunk_id'])].append(feature)

        data_list = []
        
        # --- PASS 1: COLLECT STATS ---
        all_feats = []
        temp_graphs = []

        for (lid, cid), features in grouped_data.items():
            # FILTER 1: Skip chunks with fewer than 6 nodes
            # Small chunks are topologically trivial and cause the "negative edge" warning
            if len(features) < 6:
                continue

            features.sort(key=lambda f: f['properties']['node_id'])
            
            raw = np.array([f['properties']['features'] for f in features], dtype=np.float32)
            subset = raw[:, [0, 3, 4, 5, 6]] # Area, Perim, Holes, Convexity, AR
            
            # Log transform Area & Perim
            subset[:, 0] = np.log1p(subset[:, 0]) 
            subset[:, 1] = np.log1p(subset[:, 1]) 
            
            all_feats.append(subset)
            temp_graphs.append((lid, cid, features, subset))

        if not all_feats:
            raise ValueError("All graphs were filtered out! Check your chunking logic.")

        # Calc global stats
        all_feats_concat = np.concatenate(all_feats, axis=0)
        mean = np.mean(all_feats_concat, axis=0)
        std = np.std(all_feats_concat, axis=0) + 1e-6

        # --- PASS 2: BUILD DATA ---
        skipped = 0
        for lid, cid, features, subset in temp_graphs:
            normalized_x = (subset - mean) / std
            x = torch.tensor(normalized_x, dtype=torch.float)
            
            # Position
            pos_list = []
            for f in features:
                coords = f['geometry']['coordinates'][0]
                pts = np.array(coords)
                centroid = np.mean(pts[:-1], axis=0) if len(pts) > 1 else np.mean(pts, axis=0)
                pos_list.append(centroid)
            pos = torch.tensor(np.array(pos_list), dtype=torch.float)

            # Edges
            id_map = {f['properties']['node_id']: i for i, f in enumerate(features)}
            edge_indices = []
            for i, f in enumerate(features):
                for nbr in f['properties']['neighbors']:
                    if nbr in id_map:
                        edge_indices.append([i, id_map[nbr]])
            
            if not edge_indices:
                skipped += 1
                continue
                
            edge_index = torch.tensor(edge_indices, dtype=torch.long).t().contiguous()

            # FILTER 2: Ensure enough edges exist for 80/10/10 split
            if edge_index.size(1) < 6:
                skipped += 1
                continue

            data = Data(x=x, edge_index=edge_index, pos=pos)
            data.layout_id = lid
            data.chunk_id = cid
            data_list.append(data)

        if self.pre_transform is not None:
            data_list = [self.pre_transform(d) for d in data_list]

        data, slices = self.collate(data_list)
        torch.save((data, slices), self.processed_paths[0])
        print(f"[SUCCESS] Saved {len(data_list)} graphs. Skipped {skipped} trivial chunks.")

def get_architectural_dataloaders(geojson_path, batch_size=32, split_ratio=0.8, force_reload=False):
    cache_dir = './data_cache'
    if force_reload and os.path.exists(cache_dir):
        shutil.rmtree(cache_dir)

    transform = T.Compose([
        T.RandomLinkSplit(
            num_val=0.1,  
            num_test=0.1, 
            is_undirected=True,
            add_negative_train_samples=False 
        )
    ])

    dataset = ArchitecturalGraphDataset(root=cache_dir, geojson_path=geojson_path, transform=transform)
    torch.manual_seed(42)
    dataset = dataset.shuffle()
    
    train_size = int(len(dataset) * split_ratio)
    
    train_loader = DataLoader(dataset[:train_size], batch_size=batch_size, shuffle=True)
    test_loader = DataLoader(dataset[train_size:], batch_size=batch_size, shuffle=False) if len(dataset) > train_size else None

    return train_loader, test_loader, dataset.num_features

if __name__ == "__main__":
    get_architectural_dataloaders("chunked_data.geojson", force_reload=True)
```

```python
import json
import torch
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon as MplPolygon
from matplotlib.collections import PatchCollection, LineCollection
from torch_geometric.utils import to_networkx
import networkx as nx

def build_geometry_lookup(geojson_path):
    """
    Creates a fast lookup dictionary: (layout_id, chunk_id, node_id) -> Polygon Coords
    """
    print(f"[INFO] Indexing GeoJSON geometry from {geojson_path}...")
    lookup = {}
    with open(geojson_path, 'r') as f:
        data = json.load(f)
        for feature in data['features']:
            props = feature['properties']
            lid = props['layout_id']
            cid = props['chunk_id']
            nid = props.get('node_id', props.get('chunk_node_id')) # Handle both naming conventions
            
            geom = feature['geometry']
            if geom and 'coordinates' in geom:
                # Store coordinates. Assuming simple polygon (exterior ring only for viz)
                lookup[(lid, cid, nid)] = geom['coordinates'][0]
    return lookup

def visualize_batch(batch, geojson_path, max_samples=4):
    """
    Visualizes a PyG Batch.
    - Re-calculates centroids from GeoJSON geometry to fix normalization mismatch.
    """
    # 1. Load Geometry Source
    geo_lookup = build_geometry_lookup(geojson_path)
    
    # 2. De-batch into individual Data objects
    try:
        data_list = batch.to_data_list()
    except AttributeError:
        # Fallback if a list was passed (e.g. from random_link_split)
        if isinstance(batch, list):
            data_list = batch[0].to_data_list()
        else:
            raise ValueError(f"Unexpected batch type: {type(batch)}")
            
    num_samples = min(len(data_list), max_samples)
    
    # 3. Setup Plot
    fig, axes = plt.subplots(num_samples, 2, figsize=(16, 6 * num_samples))
    if num_samples == 1: axes = [axes]
    
    print(f"[INFO] Visualizing {num_samples} samples...")

    for i in range(num_samples):
        data = data_list[i]
        lid = data.layout_id
        cid = data.chunk_id
        
        # --- A. Retrieve Geometry ---
        # Look up nodes for this chunk and sort by node_id to match Data object order
        relevant_keys = sorted(
            [k for k in geo_lookup.keys() if k[0] == lid and k[1] == cid], 
            key=lambda x: x[2]
        )
        
        polygons = []
        true_centroids = []
        
        all_coords = []
        
        for key in relevant_keys:
            poly_coords = geo_lookup[key] # List of [x, y]
            all_coords.append(poly_coords)
            
            # Create Patch
            polygons.append(MplPolygon(np.array(poly_coords), closed=True))
            
            # Calculate TRUE Centroid from geometry (ignoring data.pos)
            # Simple mean of vertices (excluding duplicate last point)
            pts = np.array(poly_coords[:-1]) 
            if len(pts) > 0:
                c = np.mean(pts, axis=0)
                true_centroids.append(c)
            else:
                true_centroids.append([0,0])

        true_centroids = np.array(true_centroids)
        
        # --- B. Plot Left: Geometry ---
        ax_geo = axes[i][0]
        if polygons:
            colors = plt.cm.tab20(np.linspace(0, 1, len(polygons)))
            pc = PatchCollection(polygons, facecolor=colors, edgecolor='black', alpha=0.7)
            ax_geo.add_collection(pc)
            
            # Scale View
            all_pts = np.concatenate([np.array(p) for p in all_coords])
            min_x, min_y = all_pts.min(axis=0)
            max_x, max_y = all_pts.max(axis=0)
            pad = max(max_x-min_x, max_y-min_y) * 0.1
            ax_geo.set_xlim(min_x - pad, max_x + pad)
            ax_geo.set_ylim(min_y - pad, max_y + pad)
            ax_geo.set_aspect('equal')
        
        ax_geo.set_title(f"Raw Geometry (Layout {lid}, Chunk {cid})")
        ax_geo.axis('off')

        # --- C. Plot Right: Graph ---
        ax_graph = axes[i][1]
        
        if len(true_centroids) > 0:
            # Draw Nodes
            ax_graph.scatter(true_centroids[:, 0], true_centroids[:, 1], c='black', s=20, zorder=5)
            
            # Draw Edges
            edge_index = data.edge_index.numpy()
            lines = []
            if edge_index.shape[1] > 0:
                for e in range(edge_index.shape[1]):
                    u, v = edge_index[0, e], edge_index[1, e]
                    # Ensure indices are within bounds (sanity check)
                    if u < len(true_centroids) and v < len(true_centroids):
                        lines.append([true_centroids[u], true_centroids[v]])
            
            if lines:
                lc = LineCollection(lines, colors='red', linewidths=1.5, alpha=0.6, zorder=10)
                ax_graph.add_collection(lc)

        # Draw Ghost Polygons
        if polygons:
            pc_ghost = PatchCollection(polygons, facecolor='#eeeeee', edgecolor='#d0d0d0', alpha=0.5, zorder=0)
            ax_graph.add_collection(pc_ghost)

        ax_graph.set_xlim(ax_geo.get_xlim())
        ax_graph.set_ylim(ax_geo.get_ylim())
        ax_graph.set_aspect('equal')
        ax_graph.set_title(f"Graph Topology (Overlaid)")
        ax_graph.axis('off')

    plt.tight_layout()
    plt.show()

# ==========================================
# EXECUTION
# ==========================================
if __name__ == "__main__":
    GEOJSON_FILE = "chunked_data.geojson"
    
    # 1. Get the loader
    train_loader, test_loader, num_features = get_architectural_dataloaders(
        GEOJSON_FILE, batch_size=4
    )
    
    # 2. Get the batch (which returns a list of 3 splits)
    batch_splits = next(iter(train_loader))
    
    # Check what we got
    print(f"Loader returned type: {type(batch_splits)}") 
    
    # 3. Extract the TRAIN batch (Index 0)
    train_batch = batch_splits[0]
    
    print(f"Visualizing Training Batch: {train_batch}")

    # 4. Visualize
    visualize_batch(train_batch, GEOJSON_FILE, max_samples=3)
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from torch_geometric.nn import GCNConv, VGAE, BatchNorm
from sklearn.decomposition import PCA
from sklearn.metrics import roc_auc_score, average_precision_score
# from dataset import get_architectural_dataloaders
import warnings

# Suppress the "not enough negative edges" warning
# warnings.filterwarnings("ignore", module="torch_geometric.transforms.random_link_split")

class VariationalGCNEncoder(torch.nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        hidden_dim = 2 * out_channels
        
        self.conv1 = GCNConv(in_channels, hidden_dim)
        self.bn1 = BatchNorm(hidden_dim) # BatchNorm helps GNNs learn faster
        
        self.conv2_mu = GCNConv(hidden_dim, out_channels)
        self.conv2_logstd = GCNConv(hidden_dim, out_channels)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = self.bn1(x) # Apply BatchNorm
        x = x.relu()
        return self.conv2_mu(x, edge_index), self.conv2_logstd(x, edge_index)

def train(model, optimizer, loader):
    model.train()
    total_loss = 0
    for batch_splits in loader:
        data = batch_splits[0]
        optimizer.zero_grad()
        z = model.encode(data.x, data.edge_index)
        loss = model.recon_loss(z, data.edge_index) + (1 / data.num_nodes) * model.kl_loss()
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(loader)

@torch.no_grad()
def test(model, loader):
    model.eval()
    auc_scores, ap_scores = [], []
    for batch_splits in loader:
        val_data = batch_splits[1]
        z = model.encode(val_data.x, val_data.edge_index)
        preds = model.decoder(z, val_data.edge_label_index, sigmoid=True)
        y_true = val_data.edge_label.cpu().numpy()
        y_pred = preds.cpu().numpy()
        try:
            auc_scores.append(roc_auc_score(y_true, y_pred))
            ap_scores.append(average_precision_score(y_true, y_pred))
        except: pass
    return np.mean(auc_scores), np.mean(ap_scores)

if __name__ == "__main__":
    GEOJSON_FILE = "chunked_data.geojson"
    EPOCHS = 100
    EMBEDDING_DIM = 32
    LR = 0.005

    print("[INFO] Loading Stricter Data...")
    train_loader, test_loader, num_features = get_architectural_dataloaders(
        GEOJSON_FILE, batch_size=16, force_reload=False
    )
    
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    encoder = VariationalGCNEncoder(num_features, EMBEDDING_DIM)
    model = VGAE(encoder).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR)
    
    print(f"[INFO] Training on {device}...")
    
    for epoch in range(1, EPOCHS + 1):
        loss = train(model, optimizer, train_loader)
        if epoch % 10 == 0:
            if test_loader:
                val_auc, val_ap = test(model, test_loader)
                print(f'Epoch: {epoch:03d}, Loss: {loss:.4f}, Val AUC: {val_auc:.4f}, Val AP: {val_ap:.4f}')
            else:
                print(f'Epoch: {epoch:03d}, Loss: {loss:.4f} (No Test Data)')

    # Visual Check
    model.eval()
    if train_loader:
        batch = next(iter(train_loader))[0]
        z = model.encode(batch.x, batch.edge_index).detach().cpu().numpy()
        
        plt.figure(figsize=(8,6))
        # Color by normalized Area (feature 0)
        plt.scatter(z[:,0], z[:,1], c=batch.x[:,0], cmap='viridis', s=50, alpha=0.8)
        plt.colorbar(label="Norm. Log(Area)")
        plt.title("Latent Space: Color should show gradient (Small -> Large Rooms)")
        plt.show()
```

# Graph masked modeling

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GATv2Conv, BatchNorm

class GraphMAE(nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels, num_layers=2, dropout=0.1, mask_rate=0.5):
        super().__init__()
        self.mask_rate = mask_rate
        
        # 1. Mask Token (Learnable "Unknown" Vector)
        self.mask_token = nn.Parameter(torch.zeros(1, in_channels))
        nn.init.xavier_normal_(self.mask_token)

        # 2. Encoder (GAT + BatchNorm)
        self.encoder_layers = nn.ModuleList()
        self.batch_norms = nn.ModuleList()
        
        # Input Layer
        self.encoder_layers.append(GATv2Conv(in_channels, hidden_channels, heads=4, concat=False))
        self.batch_norms.append(BatchNorm(hidden_channels))
        
        # Hidden Layers
        for _ in range(num_layers - 1):
            self.encoder_layers.append(GATv2Conv(hidden_channels, hidden_channels, heads=4, concat=False))
            self.batch_norms.append(BatchNorm(hidden_channels))

        self.dropout = dropout

        # 3. Decoder (MLP) - Projects back to original feature size
        self.decoder = nn.Sequential(
            nn.Linear(hidden_channels, hidden_channels),
            nn.ReLU(),
            nn.Linear(hidden_channels, out_channels)
        )

    def mask_nodes(self, x, forced_mask_rate=None):
        """
        Args:
            forced_mask_rate: If provided, overrides self.training logic.
        """
        num_nodes = x.size(0)
        
        # Determine rate: Use arg if present, else use default
        rate = forced_mask_rate if forced_mask_rate is not None else self.mask_rate
        
        # Create mask
        if rate > 0.0:
            mask = torch.rand(num_nodes, device=x.device) < rate
            # Prevent "masking everything" edge case
            if mask.all(): 
                mask[0] = False
        else:
            mask = torch.zeros(num_nodes, dtype=torch.bool, device=x.device)

        # Apply Mask Token
        x_masked = x.clone()
        x_masked[mask] = self.mask_token

        return x_masked, mask

    def encode(self, x, edge_index):
        for conv, bn in zip(self.encoder_layers, self.batch_norms):
            x = conv(x, edge_index)
            x = bn(x) # BN stabilizes the chaotic training loss
            x = F.elu(x)
            x = F.dropout(x, p=self.dropout, training=self.training)
        return x

    def forward(self, x, edge_index, forced_mask_rate=None):
        # 1. Mask
        x_masked, mask_idx = self.mask_nodes(x, forced_mask_rate=forced_mask_rate)
        
        # 2. Encode
        z = self.encode(x_masked, edge_index)
        
        # 3. Decode
        x_recon = self.decoder(z)
        
        return x_recon, mask_idx

    def loss(self, x_recon, x_target, mask_idx):
        if not mask_idx.any():
            return torch.tensor(0.0, device=x_recon.device, requires_grad=True)
        # MSE only on masked nodes
        return F.mse_loss(x_recon[mask_idx], x_target[mask_idx])
```

```python
import torch
import torch.optim as optim
import matplotlib.pyplot as plt
from torch_geometric.loader import DataLoader
import numpy as np
import os
import shutil

# ==========================================
# TRAINING HELPERS
# ==========================================

def train_one_epoch(model, loader, optimizer, device):
    model.train()
    total_loss = 0
    
    for data in loader:
        data = data.to(device)
        optimizer.zero_grad()
        
        x_target = data.x.clone()
        
        # Train: Use default mask rate (e.g. 0.5)
        x_recon, mask_idx = model(data.x, data.edge_index)
        
        loss = model.loss(x_recon, x_target, mask_idx)
        loss.backward()
        
        # Clip gradients to prevent the "1.6" loss spikes
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        
        optimizer.step()
        total_loss += loss.item()
        
    return total_loss / len(loader)

@torch.no_grad()
def evaluate(model, loader, device):
    model.eval()
    total_loss = 0
    for data in loader:
        data = data.to(device)
        x_target = data.x.clone()
        
        # FIX: Force masking during validation!
        # Otherwise validation loss is 0.0 because nothing is masked.
        x_recon, mask_idx = model(data.x, data.edge_index, forced_mask_rate=0.5)
        
        loss = model.loss(x_recon, x_target, mask_idx)
        total_loss += loss.item()
    return total_loss / len(loader)

# ==========================================
# VISUALIZATION
# ==========================================
def visualize_reconstruction(model, loader, device, feature_index=0, feature_name="Log(Area)"):
    """
    Visualizes how well the model reconstructs a specific feature (e.g., Area)
    for nodes that were masked.
    """
    model.eval()
    data = next(iter(loader))
    data = data.to(device)
    
    # Force a specific mask for visualization
    # Mask 50% of nodes
    x_input = data.x.clone()
    num_nodes = data.num_nodes
    mask = torch.rand(num_nodes, device=device) < 0.5
    
    # Apply token manually for viz
    x_input[mask] = model.mask_token
    
    # Predict
    z = model.encode(x_input, data.edge_index)
    x_recon = model.decoder(z)
    
    # Extract values for plotting
    y_true = data.x[mask, feature_index].cpu().numpy()
    y_pred = x_recon[mask, feature_index].detach().cpu().numpy()
    
    plt.figure(figsize=(12, 5))
    
    # Plot 1: Scatter correlation
    plt.subplot(1, 2, 1)
    plt.scatter(y_true, y_pred, alpha=0.6, c='blue')
    plt.plot([y_true.min(), y_true.max()], [y_true.min(), y_true.max()], 'r--')
    plt.xlabel(f"Actual {feature_name}")
    plt.ylabel(f"Predicted {feature_name}")
    plt.title(f"Reconstruction of Masked Nodes")
    
    # Plot 2: Graph View (First Sample in Batch)
    # We color nodes: Gray (Unmasked), Red (Masked & Poor Pred), Green (Masked & Good Pred)
    plt.subplot(1, 2, 2)
    
    # Unbatch first graph for clarity
    from torch_geometric.utils import to_networkx
    import networkx as nx
    
    # Isolate the first graph in the batch
    batch_mask = data.batch == 0
    sub_nodes = batch_mask.nonzero().view(-1)
    
    G = to_networkx(data, to_undirected=True)
    subgraph = G.subgraph(sub_nodes.cpu().numpy())
    subgraph = nx.convert_node_labels_to_integers(subgraph)
    
    # Get positions
    pos_data = data.pos[batch_mask].cpu().numpy()
    pos = {i: (pos_data[i][0], pos_data[i][1]) for i in range(len(pos_data))}
    
    # Determine colors
    node_colors = []
    local_mask = mask[batch_mask].cpu().numpy()
    
    for i in range(len(local_mask)):
        if not local_mask[i]:
            node_colors.append("lightgray") # Context node
        else:
            # Masked node: Color intensity by error?
            node_colors.append("red") # Masked
            
    nx.draw(subgraph, pos, node_color=node_colors, node_size=50, edge_color="gray", width=0.5)
    plt.title("Red Nodes were Masked")
    
    plt.tight_layout()
    plt.show()

# ==========================================
# MAIN EXECUTION
# ==========================================
if __name__ == "__main__":
    GEOJSON = "chunked_data.geojson"
    EPOCHS = 200
    HIDDEN_DIM = 128  # Increased capacity
    LR = 0.0005       # Lower LR for stability
    BATCH_SIZE = 16
    
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"[INFO] Device: {device}")

    # 1. Load Data (Clear cache to be safe)
    if os.path.exists('./data_cache_mgm'): shutil.rmtree('./data_cache_mgm')
    dataset = ArchitecturalGraphDataset(root='./data_cache_mgm', geojson_path=GEOJSON)
    dataset = dataset.shuffle()
    
    # Split
    train_size = int(len(dataset) * 0.85)
    train_loader = DataLoader(dataset[:train_size], batch_size=BATCH_SIZE, shuffle=True)
    test_loader = DataLoader(dataset[train_size:], batch_size=BATCH_SIZE, shuffle=False)
    
    print(f"[INFO] Train Graphs: {len(dataset[:train_size])} | Val Graphs: {len(dataset[train_size:])}")

    # 2. Model
    model = GraphMAE(
        in_channels=dataset.num_features,
        hidden_channels=HIDDEN_DIM,
        out_channels=dataset.num_features,
        mask_rate=0.50 # Mask 50% of nodes
    ).to(device)
    
    optimizer = optim.Adam(model.parameters(), lr=LR, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=15, factor=0.5)

    # 3. Train
    print("[INFO] Starting MGM Training...")
    loss_history = {'train': [], 'val': []}
    
    for epoch in range(1, EPOCHS + 1):
        train_loss = train_one_epoch(model, train_loader, optimizer, device)
        val_loss = evaluate(model, test_loader, device)
        
        scheduler.step(val_loss)
        loss_history['train'].append(train_loss)
        loss_history['val'].append(val_loss)
        
        if epoch % 10 == 0:
            current_lr = optimizer.param_groups[0]['lr']
            print(f"Epoch {epoch:03d} | Train: {train_loss:.4f} | Val: {val_loss:.4f} | LR: {current_lr:.6f}")

    # 4. Plot Loss
    plt.figure(figsize=(10,5))
    plt.plot(loss_history['train'], label='Train Loss')
    plt.plot(loss_history['val'], label='Val Loss')
    plt.title("MGM Reconstruction Loss (With BatchNorm & Gradient Clip)")
    plt.legend()
    plt.yscale('log') # Log scale helps see progress when loss is small
    plt.show()
    
    # 5. Visual Sanity Check
    visualize_reconstruction(model, test_loader, device, feature_index=0, feature_name="Norm. Log(Area)")
```
