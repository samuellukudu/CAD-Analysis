```python
"""
dxf_to_chunks_visual.py

Combines:
1. ROBUST PIPELINE: Batched processing, simplified geometry (No crashes).
2. RICH VISUALIZATION: Side-by-side Geometry + Topology plots.
3. CHUNKING: K-Means clustering for spatial partitioning.
"""

import math
import os
import json
import gc
import numpy as np
import networkx as nx
import ezdxf
import matplotlib
# CRITICAL: Use Agg backend to prevent Jupyter/Browser memory crash
matplotlib.use('Agg') 
import matplotlib.pyplot as plt

from collections import defaultdict
from sklearn.cluster import KMeans
from shapely.geometry import LineString, Point, box
from shapely.ops import unary_union, polygonize
from shapely.strtree import STRtree
from matplotlib.patches import Polygon as MplPolygon
from matplotlib.collections import PatchCollection
from matplotlib import colormaps
from tqdm import tqdm

# ==========================================
# CONFIGURATION
# ==========================================
BATCH_SIZE = 1000
ARC_RES = 6
# Only visualize the top N largest layouts to save RAM
VIZ_LIMIT = 1 

# ==========================================
# 1. Extraction (Stable)
# ==========================================
def sample_arc(center, radius, start_angle, end_angle, n_segments=6):
    if end_angle <= start_angle: end_angle += 360.0
    start, end = math.radians(start_angle), math.radians(end_angle)
    angles = np.linspace(start, end, max(3, n_segments))
    return [(center[0] + radius * math.cos(a), center[1] + radius * math.sin(a)) for a in angles]

def extract_dxf_data(dxf_path):
    if not os.path.exists(dxf_path): return []
    try:
        doc = ezdxf.readfile(dxf_path)
    except Exception as e:
        print(f"[ERROR] {e}")
        return []

    msp = doc.modelspace()
    results = []
    print(f"[INFO] Scanning entities...")

    ignore = {'TEXT', 'MTEXT', 'DIMENSION', 'HATCH', 'INSERT', 'LEADER', 'SPLINE', 'POINT'}
    
    for entity in msp:
        if entity.dxftype() in ignore: continue
        geoms = []
        try:
            if entity.dxftype() == "LINE":
                geoms.append(LineString([(entity.dxf.start.x, entity.dxf.start.y), (entity.dxf.end.x, entity.dxf.end.y)]))
            elif entity.dxftype() == "LWPOLYLINE":
                pts = list(entity.get_points('xy'))
                if len(pts) >= 2: geoms.append(LineString(pts))
            elif entity.dxftype() == "CIRCLE":
                pts = sample_arc((entity.dxf.center.x, entity.dxf.center.y), entity.dxf.radius, 0, 360, ARC_RES)
                geoms.append(LineString(pts))
            elif entity.dxftype() == "ARC":
                pts = sample_arc((entity.dxf.center.x, entity.dxf.center.y), entity.dxf.radius, entity.dxf.start_angle, entity.dxf.end_angle, ARC_RES)
                geoms.append(LineString(pts))
        except: pass
        
        for ln in geoms:
            if ln.length > 0.01:
                rounded_coords = [(round(x, 3), round(y, 3)) for x, y in ln.coords]
                results.append({
                    'geometry': LineString(rounded_coords),
                    'layer': entity.dxf.layer,
                    'handle': entity.dxf.handle
                })
    return results

# ==========================================
# 2. Robust Building & Merging
# ==========================================
def batched_union(geoms):
    if not geoms: return None
    batches = [geoms[i:i + BATCH_SIZE] for i in range(0, len(geoms), BATCH_SIZE)]
    intermediates = []
    for b in tqdm(batches, desc="Union", leave=False):
        try:
            intermediates.append(unary_union(b))
        except: continue
    return unary_union(intermediates)

def build_regions(line_data):
    if not line_data: return []
    raw_lines = [d['geometry'] for d in line_data]
    
    print(f"[INFO] Processing {len(raw_lines)} lines...")
    union_geom = batched_union(raw_lines)
    
    # Polygonize
    polys = list(polygonize(union_geom))
    if not polys:
        # Fallback: Thick lines for gap closing
        print("[WARN] Exact union failed. Trying gap closing...")
        thick_lines = [g.buffer(0.05) for g in raw_lines]
        union_geom = unary_union(thick_lines)
        polys = list(polygonize(union_geom.boundary))

    # Attribute Assignment
    line_tree = STRtree(raw_lines)
    regions = []
    
    print(f"[INFO] Attributing {len(polys)} regions...")
    for poly in polys:
        if poly.area < 0.1: continue
        simple_poly = poly.simplify(0.01)
        
        indices = line_tree.query(simple_poly)
        votes = defaultdict(float)
        
        for idx in indices:
            ln = line_data[idx]
            if simple_poly.distance(ln['geometry']) < 0.1:
                votes[ln['layer']] += 1
                
        dom_layer = max(votes, key=votes.get) if votes else "UNKNOWN"
        
        # Calculate features
        c = simple_poly.centroid
        feat = np.array([simple_poly.area, c.x, c.y], dtype=float)
        
        regions.append({
            "geometry": simple_poly,
            "layer": dom_layer,
            "features": feat,
            "raw_features": feat, # For viz compat
            "labels": []
        })
    
    del union_geom
    gc.collect()
    return regions

def merge_regions(regions):
    if not regions: return []
    print(f"[INFO] Merging layers...")
    
    geoms = [r['geometry'] for r in regions]
    tree = STRtree(geoms)
    G = nx.Graph()
    G.add_nodes_from(range(len(regions)))
    
    for i, geom in enumerate(geoms):
        bbox = geom.bounds
        search_box = box(bbox[0]-0.2, bbox[1]-0.2, bbox[2]+0.2, bbox[3]+0.2)
        hits = tree.query(search_box)
        for j in hits:
            if i < j:
                if regions[i]['layer'] == regions[j]['layer']:
                    if geom.distance(geoms[j]) < 0.2:
                        G.add_edge(i, j)

    merged = []
    for comp in nx.connected_components(G):
        indices = list(comp)
        if len(indices) == 1:
            merged.append(regions[indices[0]])
        else:
            sub_geoms = [regions[i]['geometry'] for i in indices]
            fused = unary_union(sub_geoms)
            if fused.geom_type == 'MultiPolygon':
                fused = max(fused.geoms, key=lambda a: a.area)
            
            p = regions[indices[0]]
            # Recalculate centroid
            c = fused.centroid
            feat = np.array([fused.area, c.x, c.y], dtype=float)
            
            merged.append({
                "geometry": fused,
                "layer": p['layer'],
                "features": feat,
                "raw_features": feat,
                "labels": []
            })
            
    print(f"[INFO] Merged to {len(merged)} regions.")
    gc.collect()
    return merged

# ==========================================
# 3. Graph Creation
# ==========================================
def create_global_graph(regions):
    if not regions: return nx.Graph()
    
    all_geoms = [r['geometry'] for r in regions]
    minx = min(g.bounds[0] for g in all_geoms)
    maxx = max(g.bounds[2] for g in all_geoms)
    dist = max(0.5, min(5.0, (maxx-minx)*0.01))
    
    print(f"[INFO] Linking neighbors (Dist: {dist:.2f})...")
    G = nx.Graph()
    for i, r in enumerate(regions):
        G.add_node(i, features=r['features'], raw_xy=r['raw_features'][1:3])
        
    tree = STRtree(all_geoms)
    for i, geom in enumerate(all_geoms):
        bbox = geom.bounds
        search_box = box(bbox[0]-dist, bbox[1]-dist, bbox[2]+dist, bbox[3]+dist)
        hits = tree.query(search_box)
        for j in hits:
            if i < j and geom.distance(all_geoms[j]) < dist:
                G.add_edge(i, j)
    return G

# ==========================================
# 4. Layouts & K-Means Chunking
# ==========================================
def get_layouts(global_graph, all_regions):
    components = list(nx.connected_components(global_graph))
    layouts = []
    
    # Filter small components
    valid_comps = [c for c in components if len(c) > 2]
    
    for comp in valid_comps:
        subgraph = global_graph.subgraph(comp).copy()
        
        # Relabel 0..N
        node_list = list(subgraph.nodes())
        mapping = {global_id: i for i, global_id in enumerate(node_list)}
        subgraph = nx.relabel_nodes(subgraph, mapping)
        
        layout_regions = [all_regions[global_id] for global_id in node_list]
        
        layouts.append({
            "graph": subgraph, 
            "regions": layout_regions
        })
        
    layouts.sort(key=lambda x: len(x['regions']), reverse=True)
    return layouts

def create_chunks(layout, nodes_per_chunk=30):
    graph = layout['graph']
    n_nodes = len(graph.nodes())
    if n_nodes < 2: return [graph]
    
    # 1. K-Means
    k = max(1, math.ceil(n_nodes / nodes_per_chunk))
    coords = np.array([graph.nodes[n]['raw_xy'] for n in graph.nodes()])
    
    try:
        kmeans = KMeans(n_clusters=k, n_init=5, random_state=42).fit(coords)
        labels = kmeans.labels_
    except:
        # Fallback if Singular Matrix
        return [graph]
    
    # 2. Build Chunks
    chunks = []
    for cid in range(k):
        core = [n for n, l in zip(graph.nodes(), labels) if l == cid]
        if not core: continue
        
        # 1-hop overlap
        chunk_nodes = set(core)
        for n in list(chunk_nodes):
            chunk_nodes.update(graph.neighbors(n))
            
        sub = graph.subgraph(list(chunk_nodes)).copy()
        
        # Attributes for Visualizer
        nx.set_node_attributes(sub, {n: n for n in sub.nodes()}, 'layout_node_id')
        is_core = {n: (n in core) for n in sub.nodes()}
        nx.set_node_attributes(sub, is_core, 'is_core')
        
        # Relabel for PyG (optional here, but good practice)
        # We KEEP original IDs in 'layout_node_id' for mapping back to geometry
        
        chunks.append(sub)
        
    return chunks

# ==========================================
# 5. Visualization (The Requested Style)
# ==========================================
def visualize_chunks(layouts_with_chunks, save_path):
    if not layouts_with_chunks: return
    
    # LIMIT to the top N layouts to prevent crashing
    valid_items = layouts_with_chunks[:VIZ_LIMIT]
    n_layouts = len(valid_items)
    
    print(f"[INFO] Generating Visualization for top {n_layouts} layouts...")
    
    fig, axes = plt.subplots(n_layouts, 2, figsize=(20, 10 * n_layouts), squeeze=False)
    cmap = colormaps.get_cmap("tab20")
    
    for i, item in enumerate(valid_items):
        layout_id = item['id']
        regions = item['layout_data']['regions']
        chunk_graphs = item['chunk_graphs']
        
        # --- Prepare Colors ---
        region_colors = [None] * len(regions)
        
        for c_idx, cg in enumerate(chunk_graphs):
            color = cmap(c_idx % 20)
            for u, data in cg.nodes(data=True):
                # We need the Layout-Local ID
                original_id = data.get('layout_node_id', u)
                
                if data.get('is_core', True):
                    region_colors[original_id] = color
                elif region_colors[original_id] is None:
                    # Overlap color
                    region_colors[original_id] = (0.9, 0.9, 0.9, 0.5)

        final_colors = [c if c is not None else (0.8, 0.8, 0.8, 0.3) for c in region_colors]

        # --- Plot 1: Geometry ---
        ax_geo = axes[i, 0]
        patches = []
        for r in regions:
            poly = r['geometry']
            patches.append(MplPolygon(np.array(poly.exterior.coords), closed=True))
            
        p = PatchCollection(patches, facecolor=final_colors, edgecolor='black', lw=0.5, alpha=0.9)
        ax_geo.add_collection(p)
        ax_geo.autoscale_view()
        ax_geo.set_title(f"Layout {layout_id}: Geometry (Colored by Chunk)")
        ax_geo.set_aspect('equal')
        
        # --- Plot 2: Graph ---
        ax_graph = axes[i, 1]
        G = item['layout_data']['graph']
        
        # Positions from region centroids
        pos = {n: (regions[n]['raw_features'][1], regions[n]['raw_features'][2]) for n in G.nodes()}
        
        node_c = [final_colors[n] if n < len(final_colors) else (0,0,0,1) for n in G.nodes()]
        
        # Draw edges lightly
        nx.draw_networkx_edges(G, pos, ax=ax_graph, edge_color='gray', width=0.5, alpha=0.5)
        # Draw nodes
        nx.draw_networkx_nodes(G, pos, ax=ax_graph, node_size=20, node_color=node_c)
        
        ax_graph.set_title(f"Graph Topology (Colored by Chunk)")
        ax_graph.set_aspect('equal')
        ax_graph.axis('off')

    plt.tight_layout()
    plt.savefig(save_path)
    plt.close(fig) # FREE MEMORY
    print(f"[SUCCESS] Visualization saved to {save_path}")

# ==========================================
# MAIN
# ==========================================
def pipeline(dxf_path):
    gc.collect()
    
    # 1. Extract & Build
    raw = extract_dxf_data(dxf_path)
    if not raw: return
    regions = build_regions(raw)
    
    # 2. Merge (Layers)
    regions = merge_regions(regions)
    
    # 3. Global Graph
    global_G = create_global_graph(regions)
    
    # 4. Layouts
    layouts = get_layouts(global_G, regions)
    
    # 5. Chunking & Viz Prep
    final_output = []
    print(f"[INFO] Chunking {len(layouts)} layouts...")
    
    for idx, l in enumerate(layouts):
        chunks = create_chunks(l, nodes_per_chunk=30)
        final_output.append({
            'id': idx,
            'layout_data': l,
            'chunk_graphs': chunks
        })
        
    # 6. Save & Visualize
    visualize_chunks(final_output, "chunks_viz.png")

if __name__ == "__main__":
    pipeline("data/AEC Plan Elev Sample.dxf")
```

```python
import os
import json
import torch
import numpy as np
from torch_geometric.data import InMemoryDataset, Data
from collections import defaultdict
from shapely.geometry import Polygon
import shutil

class ArchitecturalGraphDataset(InMemoryDataset):
    def __init__(self, root, source_path, transform=None, pre_transform=None, force_reprocess=False):
        """
        Args:
            root (str): Root directory where the dataset should be saved.
            source_path (str): Path to a single .geojson file OR a folder containing .geojson files.
            force_reprocess (bool): If True, deletes processed data to force a rebuild.
        """
        self.source_path = source_path
        
        # 1. Determine Input Type (File vs Folder)
        self._file_list = []
        if os.path.isdir(self.source_path):
            self._file_list = sorted([
                f for f in os.listdir(self.source_path) 
                if f.endswith('.geojson')
            ])
        elif os.path.isfile(self.source_path):
            self._file_list = [os.path.basename(self.source_path)]
        else:
            raise FileNotFoundError(f"Source path not found: {self.source_path}")

        if not self._file_list:
            raise RuntimeError(f"No .geojson files found in {self.source_path}")

        # Handle Force Reprocess Logic
        if force_reprocess:
            processed_dir = os.path.join(root, 'processed')
            if os.path.exists(processed_dir):
                print(f"[INFO] force_reprocess=True. Deleting {processed_dir} to rebuild dataset...")
                shutil.rmtree(processed_dir)

        super().__init__(root, transform, pre_transform)
        
        # Load the data (weights_only=False is required for PyG Data objects)
        self.data, self.slices = torch.load(self.processed_paths[0], weights_only=False)

    @property
    def raw_file_names(self):
        # PyG expects a list of file names in the raw_dir
        return self._file_list

    @property
    def processed_file_names(self):
        return ['architectural_graphs.pt']

    def download(self):
        """
        Copies files from the source_path (user location) to the raw_dir (PyG cache).
        """
        print(f"[INFO] Copying {len(self.raw_file_names)} files from source to raw_dir...")
        
        for file_name in self.raw_file_names:
            if os.path.isdir(self.source_path):
                src = os.path.join(self.source_path, file_name)
            else:
                src = self.source_path
                
            dst = os.path.join(self.raw_dir, file_name)
            
            # Copy if missing (or rely on force_reprocess logic to clear raw folder if needed)
            if not os.path.exists(dst):
                shutil.copy(src, dst)

    def process(self):
        print(f"[INFO] Processing {len(self.raw_paths)} files...")
        
        data_list = []
        
        # We assign a unique ID to every file being processed
        for file_idx, raw_path in enumerate(self.raw_paths):
            print(f"   -> [{file_idx+1}/{len(self.raw_paths)}] Loading {os.path.basename(raw_path)}")
            
            with open(raw_path, 'r') as f:
                data_json = json.load(f)

            # CRITICAL: Re-initialize grouping for EVERY file.
            # This ensures that Layout 1 in "File A" is NOT merged with Layout 1 in "File B".
            grouped_data = defaultdict(lambda: defaultdict(list))
            
            # 1. Parse GeoJSON
            feature_count = 0
            for feature in data_json['features']:
                feature_count += 1
                p = feature['properties']
                geom = feature['geometry']['coordinates']
                grouped_data[p['layout_id']][p['chunk_id']].append((p, geom))
                
            # 2. Build Graphs for THIS file
            for lid in sorted(grouped_data.keys()):
                for cid in sorted(grouped_data[lid].keys()):
                    
                    items = grouped_data[lid][cid]
                    # Sort by node_id to ensure X matches Adjacency
                    items.sort(key=lambda x: x[0]['node_id'])
                    
                    node_props = [i[0] for i in items]
                    raw_coords = [i[1] for i in items]

                    # Features (X)
                    features = [n['features'] for n in node_props]
                    x = torch.tensor(np.array(features), dtype=torch.float)

                    # Labels (Y)
                    if 'structure_id' in node_props[0]:
                        y = torch.tensor([n['structure_id'] for n in node_props], dtype=torch.long)
                    else:
                        y = None

                    # Edges
                    edge_list = []
                    for n in node_props:
                        u = int(n['node_id'])
                        if 'neighbors' in n:
                            for v in n['neighbors']:
                                edge_list.append([u, int(v)])

                    if edge_list:
                        edge_index = torch.tensor(edge_list, dtype=torch.long).t().contiguous()
                    else:
                        edge_index = torch.empty((2, 0), dtype=torch.long)

                    # Positional Encoding (Centroids)
                    centroids = []
                    for coords in raw_coords:
                        if len(coords) > 0:
                            poly = Polygon(coords[0])
                            centroids.append([poly.centroid.x, poly.centroid.y])
                        else:
                            centroids.append([0.0, 0.0])
                    pos = torch.tensor(np.array(centroids), dtype=torch.float)

                    data = Data(x=x, edge_index=edge_index, y=y, pos=pos)
                    
                    # Metadata
                    data.layout_id = torch.tensor([int(lid)], dtype=torch.long)
                    data.chunk_id = torch.tensor([int(cid)], dtype=torch.long)
                    
                    # [NEW] File ID: Keeps track of which file this graph came from
                    # This ensures uniqueness even if layout_ids overlap between files
                    data.file_id = torch.tensor([int(file_idx)], dtype=torch.long)

                    if self.pre_transform is not None:
                        data = self.pre_transform(data)

                    data_list.append(data)

        # 3. Save
        data, slices = self.collate(data_list)
        print(f"[SUCCESS] Created unified dataset with {len(data_list)} graphs from {len(self.raw_paths)} files.")
        torch.save((data, slices), self.processed_paths[0])

# ==========================================
# USAGE
# ==========================================
if __name__ == "__main__":
    
    # 1. SETUP: Point this to a folder OR a file
    # Example: "chunked_data" (Folder) or "chunked_data/MyFile.geojson" (File)
    ROOT_DIR = "./dataset_root"
    SOURCE_PATH = "chunked_data" 
    
    # 2. Logic to detect stale cache (Updated for folders)
    processed_file = os.path.join(ROOT_DIR, "processed", "architectural_graphs.pt")
    force_update = False
    
    # Check if processed file exists, if not force update
    if not os.path.exists(processed_file):
        force_update = True
    else:
        # If input is a directory, check if ANY file is newer than the cache
        proc_mtime = os.path.getmtime(processed_file)
        
        if os.path.isdir(SOURCE_PATH):
            for f in os.listdir(SOURCE_PATH):
                if f.endswith('.geojson'):
                    f_path = os.path.join(SOURCE_PATH, f)
                    if os.path.getmtime(f_path) > proc_mtime:
                        print(f"[WARN] {f} is newer than cache. Rebuilding...")
                        force_update = True
                        break
        elif os.path.isfile(SOURCE_PATH):
            if os.path.getmtime(SOURCE_PATH) > proc_mtime:
                print("[WARN] Source file is newer than cache. Rebuilding...")
                force_update = True

    # 3. Instantiate Dataset
    dataset = ArchitecturalGraphDataset(
        root=ROOT_DIR, 
        source_path=SOURCE_PATH, 
        force_reprocess=force_update
    )

    # 4. Verify Data
    print(f"\nDataset Info:")
    print(f"Total Graphs:   {len(dataset)}")
    
    # Check uniqueness
    # We look at (file_id, layout_id) to see unique layouts across the whole dataset
    if len(dataset) > 0:
        sample = dataset[0]
        print(f"Features:       {dataset.num_features}")
        
        # Access internal storage safely
        file_ids = dataset._data.file_id.numpy()
        layout_ids = dataset._data.layout_id.numpy()
        
        # Combine them to see distinct layouts
        unique_combinations = set(zip(file_ids, layout_ids))
        print(f"Unique Layouts (across all files): {len(unique_combinations)}")
```
