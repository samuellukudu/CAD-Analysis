```python
import ezdxf
from ezdxf.addons import odafc
```

```python
dwg_path = "AutoCAD_DWG/四清单合一/20250702-景观详图组_t3.dwg"

try:
    # Load the DWG and convert internally to DXF (returns a Drawing object)
    doc = odafc.readfile(dwg_path, version="R2018")  # Specify target DXF/ACAD version if desired
    
    # Save as a permanent DXF file if you need it on disk
    doc.saveas("output_file.dxf")
    
    print("Conversion and save successful!")
except Exception as e:
    print(f"Conversion failed: {e}")
```
